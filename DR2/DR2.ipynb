```python
IMG_SIZE = 512

def extract_green_channel(img_rgb):
    """
    Paper Section: Green Channel Extraction.
    The green channel gives the best contrast between blood vessels,
    exudates, and hemorrhages — neither under-illuminated nor over-saturated.
    Returns a 3-channel image where only the green channel is used for analysis,
    but we keep RGB shape for CNN compatibility.
    """
    green = img_rgb[:, :, 1]  # extract green channel
    # Stack into 3-channel for CNN (green replicated across channels)
    return np.stack([green, green, green], axis=-1)


def apply_clahe(img_rgb):
    """
    Paper Section: Contrast Enhancement.
    CLAHE (Contrast Limited Adaptive Histogram Equalization) divides the image
    into small blocks and equalizes each block independently.
    Applied to the green channel to enhance vessels, exudates, hemorrhages.
    """
    clahe = cv2.createCLAHE(clipLimit=2.0, tileGridSize=(8, 8))
    channels = list(cv2.split(img_rgb))
    channels[1] = clahe.apply(channels[1])  # enhance green channel only
    return cv2.merge(channels)


def crop_to_field_of_view(img, tol=7):
    gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
    mask = gray > tol
    coords = np.argwhere(mask)
    if coords.size == 0:
        return img
    y0, x0 = coords.min(axis=0)
    y1, x1 = coords.max(axis=0) + 1
    return img[y0:y1, x0:x1]   # removed re-padding — black border gone


def preprocess_image(path, return_rgb=False):
    """
    Full preprocessing pipeline from the research paper:
    1. Load image
    2. Green channel extraction (best contrast for DR features)
    3. CLAHE contrast enhancement
    4. Crop to field of view (remove black border)
    5. Resize to 512x512
    Returns float32 image normalized to [0,1].
    Set return_rgb=True to also get the original RGB for visualization.
    """
    img = cv2.imread(str(path))
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Step 1: CLAHE on original RGB (enhances green channel)
    img_clahe = apply_clahe(img_rgb)

    # Step 2: Crop to field of view
    img_cropped = crop_to_field_of_view(img_clahe)

    # Step 3: Resize to 512x512
    img_resized = cv2.resize(img_cropped, (IMG_SIZE, IMG_SIZE))

    # Step 4: Extract green channel (paper: best for DR feature visibility)
    img_green = extract_green_channel(img_resized)

    # Normalize to [0, 1]
    img_out = img_green.astype(np.float32) / 255.0

    if return_rgb:
        return img_out, img_resized  # also return colour version for viz
    return img_out


print("✅ Preprocessing functions defined (green channel + CLAHE + FOV crop + resize)")
```

```python
GLCM_PROPS = ["contrast", "dissimilarity", "homogeneity", "energy", "correlation"]

def extract_color_features(pixels_green):
    return np.array([pixels_green.mean(), pixels_green.std()])

def extract_glcm_texture(patch_green_uint8):
    if patch_green_uint8.shape[0] < 2 or patch_green_uint8.shape[1] < 2:
        return np.zeros(len(GLCM_PROPS))
    glcm = graycomatrix(patch_green_uint8, distances=[1], angles=[0],
                        levels=256, symmetric=True, normed=True)
    return np.array([graycoprops(glcm, p)[0, 0] for p in GLCM_PROPS])

def extract_shape_features(binary_mask):
    props = measure.regionprops((binary_mask).astype(int))
    if not props:
        return np.zeros(5)
    p = props[0]
    return np.array([p.area, p.perimeter, p.eccentricity, p.solidity, p.extent])

def extract_shannon_entropy(pixels_green):
    hist, _ = np.histogram(pixels_green, bins=32, range=(0, 1), density=True)
    hist = hist + 1e-10
    return np.array([-(hist * np.log2(hist)).sum()])


def build_lesion_detection_map(img_rgb_uint8):
    """
    Detects lesion pixels using LOCAL contrast deviation.
    Hemorrhages = pixels darker than their LOCAL neighborhood (not global background).
    Exudates = pixels brighter than their LOCAL neighborhood.
    This works regardless of global image brightness variation.
    """
    h, w = img_rgb_uint8.shape[:2]
    green = img_rgb_uint8[:, :, 1].astype(np.float32)
    R = img_rgb_uint8[:, :, 0].astype(np.float32)
    G = img_rgb_uint8[:, :, 1].astype(np.float32)
    B = img_rgb_uint8[:, :, 2].astype(np.float32)

    # Retina mask — exclude black border
    lum = (0.299*R + 0.587*G + 0.114*B)
    retina_mask = (lum > 20).astype(np.float32)  # uint8 scale

    # Background stats for return value
    cx, cy = w//2, h//2
    mx, my = int(w*0.30), int(h*0.30)
    patch = img_rgb_uint8[cy-my:cy+my, cx-mx:cx+mx].astype(np.float32)/255.0
    gray_p = patch.mean(axis=2)
    ret = gray_p > 0.08
    if ret.sum() < 100:
        lum_mean, lum_std = 0.38, 0.05
    else:
        lv = 0.299*patch[:,:,0][ret] + 0.587*patch[:,:,1][ret] + 0.114*patch[:,:,2][ret]
        lum_mean, lum_std = float(lv.mean()), float(lv.std())

    # ── LOCAL CONTRAST: subtract local mean (gaussian blur = local average) ──
    # This removes global brightness variation and finds LOCAL anomalies only
    sigma_local = 25  # neighborhood size — larger = more local context
    local_mean = cv2.GaussianBlur(green, (0, 0), sigma_local)
    local_std_img = cv2.GaussianBlur(
        (green - local_mean)**2, (0, 0), sigma_local)
    local_std_img = np.sqrt(local_std_img + 1e-6)

    # Normalized local deviation: positive = darker than neighborhood (hemorrhage)
    #                              negative = brighter than neighborhood (exudate)
    local_dev = (local_mean - green) / (local_std_img + 1e-6)

    # ── HEMORRHAGE MAP ────────────────────────────────────────────────────────
    # Pixels significantly darker than LOCAL surroundings
    # Use 2.0 std threshold — only true anomalies, not gradual retina variation
    hm_raw = np.clip(local_dev - 2.0, 0, None) * retina_mask

    # Morphological cleanup: remove single-pixel noise, keep blobs
    kernel_open = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (3, 3))
    hm_clean = cv2.morphologyEx(
        np.clip(hm_raw * 50, 0, 255).astype(np.uint8),
        cv2.MORPH_OPEN, kernel_open
    ).astype(np.float32) / 50.0

    # Suppress vessel edges: vessels have very high local_dev too but are
    # elongated — use closing to fill blobs then compare to distinguish
    kernel_close = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (7, 7))
    hm_filled = cv2.morphologyEx(
        np.clip(hm_clean * 50, 0, 255).astype(np.uint8),
        cv2.MORPH_CLOSE, kernel_close
    ).astype(np.float32) / 50.0

    hemorrhage_map = hm_filled * retina_mask

    # ── EXUDATE MAP ───────────────────────────────────────────────────────────
    # Pixels significantly brighter than LOCAL surroundings
    ex_raw = np.clip(-local_dev - 1.5, 0, None) * retina_mask

    # Also require yellowish color (exudates are yellow-white)
    yellow = np.clip((R - B) / (G + 1e-6) - 0.3, 0, None)
    ex_clean = cv2.morphologyEx(
        np.clip(ex_raw * 50, 0, 255).astype(np.uint8),
        cv2.MORPH_OPEN, kernel_open
    ).astype(np.float32) / 50.0

    exudate_map = ex_clean * retina_mask

    return {
        'hemorrhage_map': hemorrhage_map.astype(np.float32),
        'exudate_map':    exudate_map.astype(np.float32),
        'vessel_mask':    np.zeros((h, w), dtype=np.float32),
        'lum_mean':       lum_mean,
        'lum_std':        lum_std
    }


def score_one_superpixel(img_float, img_rgb_uint8, mask, lesion_maps=None):
    pixels_rgb   = img_rgb_uint8[mask].astype(np.float32) / 255.0
    pixels_green = pixels_rgb[:, 1]

    if pixels_green.mean() < 0.05:
        return None, 0.0

    if lesion_maps is None:
        lesion_maps = build_lesion_detection_map(img_rgb_uint8)

    lum_mean = lesion_maps['lum_mean']
    lum_std  = lesion_maps['lum_std']

    lum = (0.299*pixels_rgb[:,0].mean() +
           0.587*pixels_rgb[:,1].mean() +
           0.114*pixels_rgb[:,2].mean())

    # Skip optic disc
    if lum > lum_mean + 3.0 * lum_std:
        return None, 0.0

    hm_map = lesion_maps['hemorrhage_map']
    ex_map = lesion_maps['exudate_map']

    hm_pixels = hm_map[mask]
    ex_pixels = ex_map[mask]

    # ── KEY FIX: use strict pixel-level thresholding ──────────────────────────
    # Instead of mean/max over all superpixel pixels (which dilutes small dots),
    # count only pixels that are DEFINITELY lesion (above a high threshold).
    # A superpixel scores high if it CONTAINS lesion pixels, not if it
    # averages high across all its pixels.

    # Threshold: top 15% of hemorrhage map values = definite lesion pixels
    # With local-contrast maps, positive pixels are already sparse and meaningful
    # Use 50th percentile — every pixel above median of positives is a lesion candidate
    hm_threshold = float(np.percentile(
        hm_map[hm_map > 0], 50)) if (hm_map > 0).sum() > 10 else 0.1
    ex_threshold = float(np.percentile(
        ex_map[ex_map > 0], 50)) if (ex_map > 0).sum() > 10 else 0.1

    # Fraction of this superpixel's pixels that are definite lesion pixels
    hm_lesion_fraction = float((hm_pixels > hm_threshold).mean())
    ex_lesion_fraction = float((ex_pixels > ex_threshold).mean())

    # Peak value within superpixel (catches tiny dots)
    hm_peak = float(hm_pixels.max())
    ex_peak = float(ex_pixels.max())

    # Final score: fraction drives ranking for lesion-dense superpixels,
    # peak ensures even a single strong lesion pixel captures the superpixel
    hm_final = hm_lesion_fraction * 0.6 + max(0.0, hm_peak / (hm_threshold + 1e-6) - 1.0) * 0.4
    ex_final = ex_lesion_fraction * 0.6 + max(0.0, ex_peak / (ex_threshold + 1e-6) - 1.0) * 0.4
    hm_final = max(0.0, hm_final)
    ex_final = max(0.0, ex_final)

    # GLCM texture
    rows = np.where(mask.any(axis=1))[0]
    cols = np.where(mask.any(axis=0))[0]
    green_uint8 = img_rgb_uint8[:,:,1]
    patch_g     = green_uint8[rows[0]:rows[-1]+1, cols[0]:cols[-1]+1]

    glcm_feats    = extract_glcm_texture(patch_g)
    shape_feats   = extract_shape_features(mask)
    entropy_feats = extract_shannon_entropy(pixels_green)
    color_feats   = extract_color_features(pixels_green)

    glcm_contrast    = glcm_feats[0]
    glcm_homogeneity = glcm_feats[2]
    texture          = glcm_contrast * (1 - glcm_homogeneity)

    feature_vector = np.concatenate([
        color_feats, glcm_feats, shape_feats, entropy_feats,
        np.array([ex_final]), np.array([0.0]), np.array([hm_final])
    ])

    relevance = (
        0.60 * hm_final +
        0.30 * ex_final +
        0.07 * texture  +
        0.03 * float(entropy_feats[0]) / 10
    )

    return feature_vector, relevance


print("✅ Morphological lesion detection + superpixel scoring defined")

N_SEGMENTS = 800

def generate_superpixels(image_float):
    return slic(img_as_float(image_float), n_segments=N_SEGMENTS,
                compactness=10, sigma=1, start_label=0, channel_axis=-1)

print("✅ N_SEGMENTS and generate_superpixels defined")
```

```python
PATCH_SIZE = 64   # resize each superpixel patch to this for CNN
TOP_K      = 100   # keep top-25 highest-scoring superpixels per image

def extract_topk_patches(img_path, top_k=TOP_K):
    img_float, img_rgb = preprocess_image(img_path, return_rgb=True)
    img_float_true = img_rgb.astype(np.float32) / 255.0   # true RGB, not green-replicated
    lesion_maps    = build_lesion_detection_map(img_rgb)   # ← was missing entirely
    segs           = slic(img_float_true, n_segments=N_SEGMENTS,
                          compactness=10, sigma=1, start_label=0, channel_axis=-1)

    scored = []
    for label in np.unique(segs):
        mask = segs == label
        feat_vec, score = score_one_superpixel(img_float_true, img_rgb,
                                               mask, lesion_maps)  # ← pass maps
        if feat_vec is not None:
            scored.append((label, score, feat_vec, mask))

    # Sort by descending relevance score — paper rank order preserved
    scored.sort(key=lambda x: -x[1])

    patches = []
    for label, score, feat_vec, mask in scored[:top_k]:
        rows = np.where(mask.any(axis=1))[0]
        cols = np.where(mask.any(axis=0))[0]
        if len(rows) < 4 or len(cols) < 4:
            continue
        # Crop bounding box of this superpixel from preprocessed image
        patch = img_rgb[rows[0]:rows[-1]+1, cols[0]:cols[-1]+1]
        patch = cv2.resize(patch, (PATCH_SIZE, PATCH_SIZE))
        patches.append((patch, feat_vec, score))

    return patches


print("✅ Top-K patch extraction function defined")
```

```python
# ── SUPERPIXEL ATTENTION MASK (replaces extract_topk_patches entirely) ────────
def get_superpixel_mask(img_rgb_uint8, n_segments=800, top_k=100):
    """
    Builds a soft attention mask from superpixel relevance scores.
    Top-K superpixels get weight 1.0, rest get 0.15 (preserves context).
    FIXED: img_float passed to scorer is now the true normalized RGB,
    not the green-replicated version, so R/G/B signals are preserved.
    """
    # ── CRITICAL FIX: do NOT use extract_green_channel here.
    # score_one_superpixel needs real RGB float to compute redness correctly.
    img_float = img_rgb_uint8.astype(np.float32) / 255.0   # true RGB [0,1]
    lesion_maps = build_lesion_detection_map(img_rgb_uint8)
    segs = slic(img_float, n_segments=n_segments,
                compactness=10, sigma=1,
                start_label=0, channel_axis=-1)

    scores = {}
    for label in np.unique(segs):
        mask        = segs == label
        _, score    = score_one_superpixel(img_float, img_rgb_uint8, mask, lesion_maps)
        scores[label] = score

    # Rank superpixels, build mask
    sorted_labels = sorted(scores, key=scores.get, reverse=True)
    top_labels    = set(sorted_labels[:top_k])

    attention = np.full(segs.shape, 0.05, dtype=np.float32)  # suppress background more
    for label in top_labels:
        attention[segs == label] = 1.0                        # highlight top-K

    return attention[:, :, np.newaxis]  # (H, W, 1) — broadcast over RGB


```

```python
from multiprocessing import Pool

# ── STEP 1: Precompute all masked images once ──────────────────────────────
def compute_mask_for_row(args):
    img_path, idx = args
    try:
        img = cv2.imread(str(img_path))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        img = apply_clahe(img)
        img = crop_to_field_of_view(img)
        img = cv2.resize(img, (380, 380))
        mask = get_superpixel_mask(img)
        img_masked = (img.astype(np.float32) * mask).astype(np.uint8)
        return idx, img_masked
    except Exception as e:
        print(f"⚠ {img_path}: {e}")
        return idx, None

df["orig_idx"] = range(len(df))   # track index through train/val split
args_list = [(row["img_path"], i) for i, (_, row) in enumerate(df.iterrows())]

print("Precomputing superpixel masks (runs once, ~5-8 min)...")
precomputed = {}
with Pool(processes=4) as pool:
    for idx, img_masked in tqdm(
        pool.imap(compute_mask_for_row, args_list, chunksize=8),
        total=len(args_list), desc="Masking"
    ):
        if img_masked is not None:
            precomputed[idx] = img_masked

print(f"✅ Done — {len(precomputed)} images masked and stored in RAM")

# ── STEP 2: Dataset reads from RAM — zero cost per batch ──────────────────
IMG_SIZE = 380

train_transform = A.Compose([
    A.HorizontalFlip(p=0.5),
    A.VerticalFlip(p=0.3),
    A.RandomRotate90(p=0.5),
    A.ShiftScaleRotate(shift_limit=0.05, scale_limit=0.1, rotate_limit=15, p=0.4),
    A.ColorJitter(brightness=0.2, contrast=0.2, saturation=0.1, p=0.5),
    A.GaussianBlur(blur_limit=3, p=0.2),
    A.GaussNoise(p=0.2),
    A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    ToTensorV2()
])

val_transform = A.Compose([
    A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
    ToTensorV2()
])

class RetinopathyDataset(Dataset):
    def __init__(self, dataframe, precomputed_imgs, transform=None):
        self.df        = dataframe.reset_index(drop=True)
        self.precomp   = precomputed_imgs
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        orig_idx = self.df.iloc[idx]["orig_idx"]
        img      = self.precomp[orig_idx].copy()
        if self.transform:
            img = self.transform(image=img)["image"]
        label = torch.tensor(self.df.iloc[idx]["diagnosis"], dtype=torch.long)
        return img, label

# ── STEP 3: Split and build loaders ───────────────────────────────────────
train_df, val_df = train_test_split(
    df, test_size=0.2, stratify=df["diagnosis"], random_state=42
)

train_ds = RetinopathyDataset(train_df, precomputed, train_transform)
val_ds   = RetinopathyDataset(val_df,   precomputed, val_transform)

class_counts     = np.bincount(train_df["diagnosis"].values, minlength=5)
class_weights_np = 1.0 / (class_counts.astype(float) + 1e-6)
sample_weights   = [class_weights_np[l] for l in train_df["diagnosis"].values]

sampler = WeightedRandomSampler(
    weights=sample_weights, num_samples=len(train_df), replacement=True
)

train_loader = DataLoader(train_ds, batch_size=8, sampler=sampler,
                          num_workers=2, pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=8, shuffle=False,
                          num_workers=2, pin_memory=True)

print(f"✅ Train: {len(train_ds)} | Val: {len(val_ds)}")
print(f"Class counts: {dict(enumerate(class_counts))}")
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {device}")

def build_model(num_classes=5):
    model = efficientnet_b3(weights=EfficientNet_B3_Weights.DEFAULT)
    in_features = model.classifier[1].in_features
    model.classifier = nn.Sequential(
        nn.Dropout(p=0.4),
        nn.Linear(in_features, 256),
        nn.ReLU(),
        nn.Dropout(p=0.2),
        nn.Linear(256, num_classes)
    )
    return model

model = build_model().to(device)
print("✅ EfficientNet-B3 model ready")

# NEW: weighted loss — penalises wrong predictions on rare classes more
class_weights_tensor = torch.tensor(
    class_weights_np / class_weights_np.sum(),
    dtype=torch.float
).to(device)
criterion = nn.CrossEntropyLoss(weight=class_weights_tensor)

EPOCHS    = 30                  # was 25
optimizer = AdamW(model.parameters(), lr=1e-4, weight_decay=1e-4)

# NEW: OneCycleLR — replaces CosineAnnealingLR, converges better
scheduler = OneCycleLR(
    optimizer,
    max_lr          = 1e-3,
    steps_per_epoch = len(train_loader),
    epochs          = EPOCHS,
    pct_start       = 0.2      # 20% warmup then decay
)

history    = {"train_loss": [], "val_acc": [], "val_kappa": []}
best_kappa = 0.0               # now save by QWK not accuracy

for epoch in range(EPOCHS):

    # Training
    model.train()
    running_loss = 0.0
    for imgs, labels in train_loader:
        imgs, labels = imgs.to(device), labels.to(device)
        optimizer.zero_grad()
        outputs = model(imgs)
        loss    = criterion(outputs, labels)
        loss.backward()
        optimizer.step()
        scheduler.step()       # NEW: step every batch not every epoch
        running_loss += loss.item()

    # Validation
    model.eval()
    all_preds, all_labels = [], []
    with torch.no_grad():
        for imgs, labels in val_loader:
            imgs  = imgs.to(device)
            preds = model(imgs).argmax(dim=1).cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())

    acc      = accuracy_score(all_labels, all_preds)
    kappa    = cohen_kappa_score(all_labels, all_preds, weights="quadratic")  # NEW
    loss_avg = running_loss / len(train_loader)

    history["train_loss"].append(loss_avg)
    history["val_acc"].append(acc)
    history["val_kappa"].append(kappa)   # NEW

    if kappa > best_kappa:               # save by kappa not accuracy
        best_kappa = kappa
        torch.save(model.state_dict(), "/content/best_model.pth")

    print(f"Epoch {epoch+1:02d}/{EPOCHS} | "
          f"Loss: {loss_avg:.4f} | "
          f"Acc: {acc:.4f} | "
          f"QWK: {kappa:.4f} | "
          f"Best QWK: {best_kappa:.4f}")

print(f"\n✅ Training complete. Best QWK: {best_kappa:.4f}")
```

```python
def predict_single_image(img_path, model):
    model.eval()
    img = cv2.imread(str(img_path))
    img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    img = apply_clahe(img)
    img = crop_to_field_of_view(img)
    img = cv2.resize(img, (380, 380))

    mask = get_superpixel_mask(img)
    img  = (img.astype(np.float32) * mask).astype(np.uint8)

    tensor = val_transform(image=img)["image"].unsqueeze(0).to(device)
    with torch.no_grad():
        probs = torch.softmax(model(tensor), dim=1).cpu().numpy()[0]

    pred = int(probs.argmax())
    print(f"Confidence    : {probs[pred]*100:.1f}%")
    print(f"All probs     : { {GRADE_LABELS[i]: f'{p*100:.1f}%' for i,p in enumerate(probs)} }")
    return pred


# Test on a sample
test_row  = df.sample(1, random_state=7).iloc[0]
pred      = predict_single_image(test_row["img_path"], model)
true      = test_row["diagnosis"]
_, img_rgb = preprocess_image(test_row["img_path"], return_rgb=True)

plt.figure(figsize=(5, 5))
plt.imshow(img_rgb)
plt.title(
    f"True:      Grade {true} — {GRADE_LABELS[true]}\n"
    f"Predicted: Grade {pred} — {GRADE_LABELS[pred]}\n"
    f"{'✅ Correct' if pred == true else '❌ Wrong'}"
)
plt.axis("off")
plt.show()
```
